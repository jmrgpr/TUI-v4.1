## 1. Importar librerías necesarias / Import required libraries

Usaremos Pandas, Matplotlib, Seaborn y Plotly para visualizar los datos.

We will use Pandas, Matplotlib, Seaborn and Plotly to visualize the data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import os

# Configuración de estilo
sns.set(style="whitegrid")
```

## 2. Cargar datos desde los archivos CSV / Load data from CSV files

Los resultados de los experimentos están en la carpeta `sim/results/`. Selecciona un archivo para visualizar.

The experiment results are in the `sim/results/` folder. Select a file to visualize.

```python
# Listar archivos CSV disponibles
csv_dir = '../sim/results/'
archivos = [f for f in os.listdir(csv_dir) if f.endswith('.csv')]
print('Archivos disponibles:', archivos)

# Selecciona uno para cargar
archivo = archivos[0] if archivos else None
if archivo:
    df = pd.read_csv(os.path.join(csv_dir, archivo))
    print(f"Cargado: {archivo}")
    display(df.head())
else:
    print("No se encontraron archivos CSV.")
```

## 3. Graficar datos principales / Plot main data

Ejemplo: Gráfico de línea de la recompensa por episodio.

Example: Line plot of reward per episode.

```python
if archivo:
    # Suponiendo que hay columnas 'episode' y 'reward'
    if 'episode' in df.columns and 'reward' in df.columns:
        plt.figure(figsize=(10,5))
        sns.lineplot(data=df, x='episode', y='reward')
        plt.title('Recompensa por episodio / Reward per episode')
        plt.xlabel('Episodio / Episode')
        plt.ylabel('Recompensa / Reward')
        plt.show()
    else:
        print("No se encontraron las columnas 'episode' y 'reward'.")
```

### Ejemplo adicional: gráfico de barras / Additional example: bar plot

Puedes graficar otras columnas, por ejemplo, la frecuencia de acciones.

You can plot other columns, e.g., action frequency.

```python
if archivo and 'action' in df.columns:
    plt.figure(figsize=(8,4))
    df['action'].value_counts().plot(kind='bar')
    plt.title('Frecuencia de acciones / Action frequency')
    plt.xlabel('Acción / Action')
    plt.ylabel('Frecuencia / Frequency')
    plt.show()
```

### Ejemplo con Plotly: gráfico interactivo / Interactive plot example

Plotly permite gráficos interactivos para explorar los datos.

Plotly enables interactive plots to explore the data.

```python
if archivo and 'episode' in df.columns and 'reward' in df.columns:
    fig = px.line(df, x='episode', y='reward', title='Recompensa por episodio (Plotly)')
    fig.show()
```
