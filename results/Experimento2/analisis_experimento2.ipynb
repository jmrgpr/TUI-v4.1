````python
# Comparación de métricas relevantes entre tui y otros agentes (corregido y documentado)
# Para TUI se usan las columnas específicas, para los demás agentes las genéricas.
# Esto evita NaN y asegura que TUI aparezca correctamente.

agentes = ['control', 'dqn_control', 'simbiosis', 'tui']
metricas_por_agente = {
    'control': ['avg_pgf_neto', 'avg_reward', 'avg_tripwire'],
    'dqn_control': ['avg_pgf_neto', 'avg_reward', 'avg_tripwire'],
    'simbiosis': ['avg_pgf_neto', 'avg_reward', 'avg_tripwire'],
    'tui': ['Recompensa_tui', 'Recompensa_tui', 'Tripwires_tui']  # PGF y recompensa para TUI
}
for agente in agentes:
    print(f'\nAgente: {agente}')
    metricas = metricas_por_agente[agente]
    df_agente = df[df['agent'] == agente][metricas]
    print(df_agente.describe())

print("\nNota: Para TUI se usan las columnas 'Recompensa_tui' y 'Tripwires_tui' en vez de las genéricas para evitar NaN y mostrar los datos reales.")
````

````python
# Comparación de métricas específicas por agente (corregido y documentado)
# Se usan las columnas correctas para cada agente, incluyendo TUI.
# Así se evitan NaN y se muestran los datos reales de cada agente.

metricas_por_agente = {
    'control': ['Recompensa_control', 'Tripwires_control', 'PGF_Bruto_Avg', 'PGF_Costo_Avg'],
    'dqn_control': ['Recompensa_dqn_control', 'Tripwires_dqn_control'],
    'simbiosis': ['Recompensa_simbiosis', 'Tripwires_simbiosis'],
    'tui': ['Recompensa_tui', 'Tripwires_tui']  # Solo columnas válidas para TUI
}
for agente, metricas in metricas_por_agente.items():
    print(f'\nAgente: {agente}')
    df_agente = df[df['agent'] == agente][metricas]
    print(df_agente.describe())

print("\nNota: Para TUI solo se usan las columnas con datos válidos, evitando NaN y mostrando métricas reales.")
````

````python
# Visualización comparativa de métricas clave por agente (mejorada)
import matplotlib.pyplot as plt
import seaborn as sns
metricas_plot = {
    'control': 'Recompensa_control',
    'dqn_control': 'Recompensa_dqn_control',
    'simbiosis': 'Recompensa_simbiosis',
    'tui': 'avg_reward'
}
plt.figure(figsize=(12,7))
colores = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']
estilos = ['-', '--', '-.', ':']
for i, (agente, metrica) in enumerate(metricas_plot.items()):
    datos = df[df['agent'] == agente][metrica].dropna()
    if not datos.empty:
        sns.kdeplot(datos, label=agente, color=colores[i], linestyle=estilos[i], alpha=0.7, bw_adjust=1.2)
plt.title('Distribución de recompensa por agente (KDE mejorado)')
plt.xlabel('Recompensa')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# Visualización por agente: histogramas y estadísticas
import matplotlib.pyplot as plt
import seaborn as sns
metricas_plot = {
    'control': 'Recompensa_control',
    'dqn_control': 'Recompensa_dqn_control',
    'simbiosis': 'Recompensa_simbiosis',
    'tui': 'avg_reward'
}
fig, axes = plt.subplots(2, 2, figsize=(14,10))
axes = axes.flatten()
for i, (agente, metrica) in enumerate(metricas_plot.items()):
    datos = df[df['agent'] == agente][metrica].dropna()
    ax = axes[i]
    if not datos.empty:
        sns.histplot(datos, bins=20, kde=True, ax=ax, color='skyblue', alpha=0.7)
        media = datos.mean()
        mediana = datos.median()
        ax.axvline(media, color='red', linestyle='--', label=f'Media: {media:.2f}')
        ax.axvline(mediana, color='green', linestyle=':', label=f'Mediana: {mediana:.2f}')
        ax.set_title(f'{agente} ({metrica})')
        ax.set_xlabel('Recompensa')
        ax.set_ylabel('Frecuencia')
        ax.legend()
    else:
        ax.set_title(f'{agente} (sin datos)')
        ax.axis('off')
plt.tight_layout()
plt.show()
````

Toda la integración, ejecución y resultados de los agentes SOTA (PPO, A2C, DQN) está documentada en el archivo `DOCUMENTACION_SOTA_EXPERIMENTO2.md` y resumida en `DOCUMENTACION_CAMBIOS_EXPERIMENTO2.md`. Se recomienda consultar estos archivos para trazabilidad y detalles técnicos.

````python
# Gráficos comparativos: TUI vs SOTA (PPO, A2C, DQN) (corregido y documentado)
# Se usa Recompensa_tui para TUI en ambas comparaciones, evitando NaN y mostrando los datos reales.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob

# Cargar resultados SOTA
sota_files = glob.glob(r'c:/Proyectos/TUI-v4.1/results/Experimento2/sota_*_risk*_summary.csv')
sota_dfs = []
for f in sota_files:
    df_sota = pd.read_csv(f, header=None)
    df_sota.columns = ['riesgo', 'agent', 'avg_reward', 'avg_pgf_neto', 'robustez', 'flexibilidad', 'episodios']
    sota_dfs.append(df_sota)
sota_df = pd.concat(sota_dfs, ignore_index=True)

# Comparar recompensa promedio por riesgo
plt.figure(figsize=(10,6))
sns.lineplot(data=sota_df, x='riesgo', y='avg_reward', hue='agent', marker='o', palette='Set2')
if 'risk_scale' in df.columns:
    tui_grouped = df[df['agent'].str.strip().str.lower() == 'tui'].groupby('risk_scale')['Recompensa_tui'].mean().reset_index()
    sns.lineplot(data=tui_grouped, x='risk_scale', y='Recompensa_tui', color='#d62728', marker='o', label='TUI')
else:
    print('La columna risk_scale no existe en el DataFrame principal.')
plt.title('Comparación de recompensa promedio: TUI vs SOTA (columna corregida)')
plt.xlabel('Riesgo')
plt.ylabel('Recompensa promedio')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Comparar PGF promedio por riesgo
plt.figure(figsize=(10,6))
sns.lineplot(data=sota_df, x='riesgo', y='avg_pgf_neto', hue='agent', marker='o', palette='Set2')
if 'risk_scale' in df.columns:
    tui_grouped_pgf = df[df['agent'].str.strip().str.lower() == 'tui'].groupby('risk_scale')['Recompensa_tui'].mean().reset_index()
    sns.lineplot(data=tui_grouped_pgf, x='risk_scale', y='Recompensa_tui', color='#d62728', marker='o', label='TUI')
else:
    print('La columna risk_scale no existe en el DataFrame principal.')
plt.title('Comparación de PGF promedio: TUI vs SOTA (columna corregida)')
plt.xlabel('Riesgo')
plt.ylabel('PGF promedio')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nNota: Para TUI se usa 'Recompensa_tui' como PGF y recompensa en las comparaciones, mostrando los datos reales y evitando NaN.")
````
