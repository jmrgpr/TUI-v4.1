Carga de datos desde la carpeta reestructurada, filtros a episodios=1000, conteos y visualizaci?n b?sica.


```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Cargar master combinado (preferido) o limpio
candidatos = [Path('..')/'data'/'master_results_combined.csv', Path('..')/'data'/'master_results_clean.csv']
for c in candidatos:
    if c.exists():
        df = pd.read_csv(c)
        print('Archivo cargado:', c)
        break
else:
    raise FileNotFoundError('No se encontr? master_results_combined.csv ni master_results_clean.csv en ../data')

# Filtrar solo runs full (episodios=1000)
if 'episodes' in df.columns:
    df = df[df['episodes'] == 1000]

print('Columnas:', df.columns.tolist())
print('Primeras filas:')
display(df.head())
```

```

# Conteos por agente y riesgo
if 'agent' in df.columns and 'risk_scale' in df.columns:
    print('
Conteo por agente y riesgo:')
    print(df.groupby(['agent','risk_scale']).size())
else:
    print('No se pueden agrupar por agent/risk_scale.')
```

```

# Resumen de recompensas y tripwires por agente/riesgo
if {'agent','risk_scale','reward','tripwires'}.issubset(df.columns):
    resumen = df.groupby(['agent','risk_scale'])[['reward','tripwires']].agg(['mean','std','count']).round(3)
    display(resumen.head(20))
else:
    print('Faltan columnas reward/tripwires para el resumen.')
```

```

# Gr?fico opcional: boxplot de recompensas por agente
if {'agent','reward'}.issubset(df.columns):
    plt.figure(figsize=(8,5))
    sns.boxplot(x='agent', y='reward', data=df)
    plt.title('Distribuci?n de recompensas por agente')
    plt.show()
else:
    print('No se puede graficar; faltan columnas agent/reward.')
```
